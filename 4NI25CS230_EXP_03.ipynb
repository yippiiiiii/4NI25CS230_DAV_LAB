```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/student_data (2).csv")
print("Dataset shape:",df.shape)
```

```python
print(df.head)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["G3"],bins=20,kde=True,color="pink")
plt.title("Distribution of Final Grade")
plt.xlabel=("Final Grade")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x= "studytime", y="G3", data=df,palette= "Set2")
plt.title("Final Grade across Study Time Levels")
plt.xlabel=("Study times 91 =low,4=high")
plt.ylabel("Final Grade")
plt.show()
```

```python
sns.pairplot(df[["age","studytime","G1","G2","G3","sex"]],hue="sex",palette="husl")
plt.suptitle("Pairwise Relationships",y=1.02)
plt.show()
```

```python
numeric_cols = ["age","studytime","failures","absences","G1","G2","G3"]
plt.figure(figsize=(8,6))
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```
