```python
import pandas as pd
import numpy as np
```

```python
data={
    'Name':['Amit','Bebo','Chetan','Amit','Divya','Ravi','Sneha','Karan',
    'Pooja','Manish','Anjali','Suresh','Kavya','Rahul','Meera','Vikram',
            'Nisha','Arjun','Divya','Rohit','Swathi','Naveen','Priya','Karan','Deepa'],
    'Age':[25,np.nan,30,25,28,22,24,np.nan,27,26,23,29,np.nan,
           25,22,28,24,np.nan,28,26,23,30,27,np.nan,25],
    'Marks':[85,90,np.nan,58,78,95,88,72,np.nan,91,67,84,79,
             np.nan,93,88,76,82,78,np.nan,90,85,71,72,89],
    'City':['Mysuru','Mysuru','Bengaluru','Mysuru',np.nan,'Delhi','Mysusru',
            'Mangalore','Bengaluru',np.nan,'Mysuru','Hubli',np.nan,'Mangalore',
            'Delhi','Bengaluru',np.nan,'Mysuru','Hubli',np.nan,'Delhi',
            'Mangalore','Mysuru','Bengaluru','Mangalore']
}
df=pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\nShape of dataset (rows,columns):",df.shape)
```

```python
print("\nColumn names:",df.columns.tolist())
```

```python
print("\nData types of each column:")
print(df.dtypes)
```

```python
print("\nDataset info:")
df.info()
```

```python
print("\nDescriptive statistics (numerical columns):")
print(df.describe)
```

```python
print("\nMedian:\n",df.median(numeric_only=True))
```

```python
print("\nStandard Deviation :\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in dataset:",df.isnull().sum().sum())
```

```python
print("\nDuplicate rows in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicate rows:",df.duplicated().sum())
print("\nEDA process completed successfully")
```
